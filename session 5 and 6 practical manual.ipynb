```python
#program 1 (simple if/elif/else statement to calculate grade based on percentage)
marks = float(input("Enter your marks out of 100: "))
total_marks = int(input("Enter total marks: "))
percentage = (marks / total_marks) * 100
if percentage >= 90:
    grade = "A"
elif percentage >= 80:
    grade = "B"
elif percentage >= 70:
    grade = "C"
elif percentage >= 60:
    grade = "D"
else:
    grade = "F"
print(f"You got: {marks} out of {total_marks}")
print(f"your percentage is: {percentage:.2f}%")
print("Grade: ", grade)

```

```python
#modified program 1
subjects = ["English", "Physics", "Chemistry", "Mathematics", "Computer Science"]

total = 0

for subject in subjects:
    marks = float(input("Enter marks in " + subject + ": "))
    while marks < 1 or marks > 100:
        print("Invalid marks! Please enter marks between 1 and 100.")
        marks = float(input("Enter marks in " + subject + ": "))
    total = total + marks

percent= (total / (len(subjects) * 100) * 100)
percentage = round(percent, 2)

if percentage >= 90:
    grade = "A"
elif percentage >= 80:
    grade = "B"
elif percentage >= 70:
    grade = "C"
elif percentage >= 60:
    grade = "D"
else:
    grade = "F"
print("Total Marks =", total, "out of", len(subjects) * 100)
print("Percentage =", percentage)
print("Grade =", grade)
```

```python
#program 2(nested if--checking sign and parity together)
number = int(input("Enter a number: "))
if number > 0:
    if number % 2 == 0:
        print(f"{number} is a positive even number.")
    else:
        print(f"{number} is a positive odd number.")
elif number < 0:
    if number % 2 == 0:
        print(f"{number} is a negative even number.")
    else:
        print(f"{number} is a negative odd number.")
else:
    print(f"{number} is zero, which is neither positive nor negative.")
  
```

```python
#program 3(while loop -- sum of digits)
num = int(input("Enter a number: "))
original = num
total = 0
while num > 0:
    digit = num % 10
    total += digit
    num //= 10
print(f"The sum of digits in {original} is {total}.")
```

```python
#program 4 (While loop -- number guessing game)
secret = 7
while True:
    guess = int(input("Guess the secret number (between 1 and 10): "))
    if guess < 1 or guess > 10:
        print("Please guess a number between 1 and 10.")
        continue
    if guess < secret:
        print("Too low! Try again.")
    elif guess > secret:
        print("Too high! Try again.")
    else:
        print("Congratulations! You guessed the secret number.")
        break
```
