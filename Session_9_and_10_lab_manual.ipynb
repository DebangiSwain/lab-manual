```python
# Program 1: Defining and calling a simple function
def calculate_area(length, breadth):
  return length * breadth


print(calculate_area(5, 3))
print(calculate_area(10, 2))
```

```python
# Program 2: Default parameters and keyword arguments
def greet(name, greeting = "Hello"):
  print(greeting, name)

greet("Rhea")
greet("Kabir", "Welcome")
greet(greeting = "hi", name = "Zoya")
```

```python
# Program 3: Variable length argument using *args
def get_min_max(numbers):
  return min(numbers), max(numbers)

lowest, highest = get_min_max([4, 8, 1, 9, 3])
print("Lowest:", lowest, "Highest:", highest)
```

```python
# Program 4: Scope of a function -- local vs global
def factorial(n):
  if n == 0 or n == 1:
    return 1
  return n * factorial(n - 1)

print(factorial(5))
```

```python
# Q.1. Write a function is_prime(n) that returns true or false
def is_prime(n):
  if n <= 1:
    return False
  for i in range(2, n):
    if n % i == 0:
      return False
  return True

print(is_prime(7))
```

```python
# Q.2. Write a function that accepts any number of arguments and returns their average using *args
def average(*args):
  return sum(args) / len(args)

print(average(1, 2, 3, 4, 5))
```

```python
# Program 1: Call by value behavious (immutable arguments)
def try_to_change(n):
  n = n + 100
  return n

x = 5
result = try_to_change(x)
print("Original x:", x)
print("Returned result:", result)
```

```python
# Program 2: Call by reference behaviour (mutable arguments)
def add_item(my_list):
  my_list.append("new item")

items = ["pen", "book"]
add_item(items)
print(items) # original list WAS modified
```

```python
# Q.1. Write a function that takes a list of prices and applies a 10% discount to each price IN PLACE (modifying the original list).
def discount_prices(prices):
    for i in range(len(prices)):
        prices[i] = prices[i] * 0.9

prices = [100, 200, 300, 400]
discount_prices(prices)
print(prices)
```

```python
# Q.2. Write a recursive function to compute the sum of the first n natural numbers.
def natural_sum(n):
  if n == 1:
    return 1
  return n + natural_sum(n - 1)

print(natural_sum(5))
```
