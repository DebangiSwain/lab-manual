```python
#Program 1: for loop with Range ()
for i in range(1, 6):
  print(i, "->", i**2)
```

```python
#Program 2: range() with a step value
for i in range(0, 20, 5):
  print(i, end=" ")
print()
for i in range(10, 0, -2):
  print(i, end=" ")
```

```python
#Program 3: for loop over a list and a string
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
  print(fruit, end=" ")
print()
for letter in "Python":
  print(letter)
```

```python
#Program 4: nested for loop -- pattern printing
rows  = 5
for i in range (1, rows + 1):
  for j in range (1, i + 1):
    print("*", end=" ")
  print()
```

```python
#Q.2. Write a program to print a multiplication table (1 to 10) for a number using a for loop.
a = int(input("Enter which number's multiplication table do you want "))
b = int(input("Enter till which number multiplication table do you want "))
for i in range(1, b+1):
  print(f"{a} x {i} = {a*i}")
```

```python
#Program 1: break -- stop a loop early
numbers = [4, 8, 15, 16, 23, 42]
target = 16
for num in numbers:
  if num == target:
    print(f"Found {target} in the list.")
    break
  print(f"Checked {num},")
```

```python
#Program 2: continue -- skip an iteration
for i in range(1, 11):
  if i % 2 == 0:
    continue #skips even numbers
  print(i, end=" ")
```

```python
#Program 3: pass -- placeholder statement
for i in range (1, 6):
  if i == 3:
    pass #nothing happens, placeholder for future code
  print(i)
```

```python
#Program 4: combined example -- password attempt tracker
correct_password = "Python123#"
for attempt in range(1, 4):
  entered = input(f"Attempt {attempt}: Enter password: ")
  if entered == correct_password:
    print("Access granted!")
    break
else:
  print("Access denied!")
```

```python
#Q.1. Write a progam that prints numbers from 1 to 50 but stops as soon as it finds a number divisible by both 7 and 5.
for i in range(1, 51):
  if i % 7 == 0 and i % 5 == 0:
    print(i)
    break
  print(i, end=" ")
```

```python
#Q.2. Write a program that prints all numbers from 1 to 30, skipping multiples of 3, using continue.
for i in range(1, 31):
    if i % 3 == 0:
        continue
    print(i, end=" ")
```
