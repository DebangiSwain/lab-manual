```python
#Program 1 (List creation, indexing and slicing)
fruits = ["apple", "banana", "cherry", "mango"] #lists are mutable and indexed
print(fruits)
print(fruits[0]) #indexing starts from 0
print(fruits[-1])
print(fruits[1:3]) #prints second to third element
print(fruits[::-1]) #x:y:z, starts from x, ends at y, order is z
```

```python
#Program 2 (Common list methods)
numbers = [5, 2, 9, 1]
numbers.append(20) #adds 20 at the end
numbers.insert(1, 100) #inserts 100 at index position 1
numbers.sort() #sorts
print(numbers)
numbers.remove(100) #removes 100
print(numbers)
print("Length:", len(numbers)) #number of elements in the set
```

```python
#Program 3 (Tuple creation and immutability)
point = (10, 20)
print(point[0], point[1])
#tuples are immutable and indexed
coordinates = (12.9, 77.6)
lat, lon = coordinates #tuple unpacking
print(lat, lon)
```

```python
#Program 4 (List vs tuple -- a quick comparision)
list = [1, 2, 3]
tuple = (1, 2, 3)

list[0] = 99 #allowed, not allowed for tuple
print(f"list after change: {list} \ntuple stays same: {tuple}") 
```
