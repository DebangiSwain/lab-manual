```python
#Program 1 (set creations and operations)
A = {1, 2, 3, 4} #sets are mutable and not indexed
B = {3, 4, 5, 6}

print("Union: ", A|B)
print("Intersection: ", A&B)
print("Difference: ", A-B)

A.add(10)
print("after adding in A: ", A)
```

```python
#Program 2 (Removing duplicates using a set)
numbers = [4, 8, 4, 6, 8, 2, 6, 1]
unique_numbers = list(set(numbers)) #typecasting list as set to remove duplicates
print("Original: ", numbers)
print("Modified: ", sorted(unique_numbers))
```

```python
#Q2. Write a program that stores 5 students and their marks in a dictionary and prints the topper.
students = {"Ria": 78, "Arush": 92, "Prisha": 75, "Meghna": 96, "Ananya": 93}
topper = ""
highest_marks = 0
for student in students:
    if students[student] > highest_marks:
        highest_marks = students[student]
        topper = student
print(f"Topper: {topper} \nMarks: {highest_marks}")
```

```python
#Program 1 (String indexing, slicing and basic operations)
word = "PYTHON"
print(word[0], word[-1]) #prints first and last letter
print(word[1:4]) #prints 2,3,4th letter
print(word[::-1]) #prints word in reversed direction
print(len(word)) #prints the number of letters in the word
```

```python
#Program 4 (Palindrome check (combining string + input/output))
text = input("Enter a word: ").lower().replace(" ", "")
if text == text[::-1]:
    print(f"{text} IS a palindrome, as it is {text}")
else:
    print(f"{text} is NOT a palindrom, as it is {text[::-1]}")
```
